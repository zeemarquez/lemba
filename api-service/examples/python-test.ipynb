```python
import json, requests, pathlib, os, io
```

```python
mdpath = r'sample.md'
templatepath = r'templates\Esferize.mdt'
outputfolder = r'output'
```

```python
API_BASE_URL = 'https://lemba-az2e.onrender.com' 

md_path = pathlib.Path(mdpath)
font_path = pathlib.Path(r"fonts\Montserrat.ttf")  # local file
template_path = pathlib.Path(templatepath)

with md_path.open("rb") as md_f, font_path.open("rb") as font_f, template_path.open("rb") as tpl_f:
    resp = requests.post(
        f"{API_BASE_URL}/v1/convert/multipart",
        files=[
            ("markdown", (md_path.name, md_f, "text/markdown")),
            ("template", (template_path.name, tpl_f, "application/json")),
            ("fontFiles", ("Montserrat", font_f, "application/octet-stream")),
        ],
        data={
            "title": "My doc",
            "variables": json.dumps({"author": "Jane"}),
            # optional: remote fonts in addition to uploads
            # "fonts": json.dumps([{"family": "Inter", "url": "https://..."}]),
        },
        headers={},  # optional: {"Authorization": "Bearer YOUR_API_KEY"}
        timeout=180,
    )
resp.raise_for_status()
pathlib.Path(os.path.join(outputfolder, "out.pdf")).write_bytes(resp.content)
```
