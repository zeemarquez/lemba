```python
import json, requests, pathlib, os, io
```

```python
outputfolder = r'output'

API_BASE_URL = 'https://lemba-az2e.onrender.com'

md_path = pathlib.Path(r'sample.md')
font_path = pathlib.Path(r"fonts\Montserrat.ttf")  # local file (any filename)
template_path = pathlib.Path(r'templates\Esferize.mdt')
```

```python
with md_path.open("rb") as md_f, font_path.open("rb") as font_f, template_path.open("rb") as tpl_f:
    resp = requests.post(
        f"{API_BASE_URL}/v1/convert/multipart",
        files=[
            ("markdown", (md_path.name, md_f, "text/markdown")),
            ("template", (template_path.name, tpl_f, "application/json")),
            ("fontFiles", (font_path.name, font_f, "application/octet-stream")),
        ],
        data={
            "title": "My doc",
            "variables": json.dumps({"author": "Jane"}),
        },
        headers={},  # optional: {"Authorization": "Bearer YOUR_API_KEY"}
        timeout=180,
    )
resp.raise_for_status()
pathlib.Path(os.path.join(outputfolder, "out.pdf")).write_bytes(resp.content)
```

```python
font_url = 'https://cdn.jsdelivr.net/fontsource/fonts/montserrat@latest/latin-400-normal.ttf'

with md_path.open("rb") as md_f, template_path.open("rb") as tpl_f:
    resp = requests.post(
        f"{API_BASE_URL}/v1/convert/multipart",
        files=[
                ("markdown", (md_path.name, md_f, "text/markdown")),
                ("template", (template_path.name, tpl_f, "application/json")),
        ],
        data={
            "fonts": json.dumps(
                [{"family": "Montserrat", "url": font_url}]
            ),
        },
        timeout=180,
    )

resp.raise_for_status()
pathlib.Path(os.path.join(outputfolder, "out.pdf")).write_bytes(resp.content)
```
