```python
import json, requests, pathlib, os
```

```python
mdpath = r'sample.md'
templatepath = r'templates\Basic.mdt'
outputfolder = r'output'
```

```python
API_BASE_URL = 'https://lemba-az2e.onrender.com' 

resp = requests.post(
    f"{API_BASE_URL}/v1/convert",
    json={
        "markdown":  pathlib.Path(mdpath).read_text(encoding="utf-8"),
        "template":  json.loads(pathlib.Path(templatepath).read_text(encoding="utf-8"))
    },
    timeout=120,
)
resp.raise_for_status()

pathlib.Path(os.path.join(outputfolder, "out.pdf")).write_bytes(resp.content)
```
